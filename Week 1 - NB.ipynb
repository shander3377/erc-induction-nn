```python
# Q2) Array Indexing and Slicing Create a 6x6 matrix with values from 1 to 36. Extract and print:

#     The third row

#     The last two columns

#     All elements greater than 20

#     Every second element in the diagonal

#     Replace all values in the first and last rows with -1
p=np.arange(1,37).reshape(6,6)
print(p[2, :])
print(p[:, -2:])
print(p[p>20])
print(np.diag(p)[::2])
p[0,:]=-1
p[-1,:]=-1
print(p)
```

```python
# Q3) Mathematical Operations

#     Create two 1D arrays: a = [1, 2, 3, 4, 5] and b = [2, 4, 6, 8, 10]

#     Perform and display results for:

#     i) Element-wise addition, subtraction, multiplication, and division

#     ii) Dot product of the arrays

#     iii) Square root of array a iv) Natural logarithm of array b

p1=np.array([1,2,3,4,5])
p2=np.array([2,4,6,8,10])
#     i) Element-wise addition, subtraction, multiplication, and division
print(p1+p2)
print(p1-p2)
print(p1*p2)
print(p1/p2)

#     ii) Dot product of the arrays
print(np.dot(p1,p2))

#     iii) Square root of array a
print(np.sqrt(p1))

# iv) Natural logarithm of array b
print(np.log(p2))
```
